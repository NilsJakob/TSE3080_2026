# TSE3080 Electrical Machines

## Why english?
   The Norwegian Agency for Quality Assurance (Nasjonalt organ for kvalitet i utdanningen) NOKUT study in 2007
   Improve internationalisation of engineering courses

 - English text books ($\rightarrow$ passive learning)
  - Courses partly in English ($\rightarrow$ active learning)
   - Whole English semester
 
Benefits:

-  Prepared for study/work abroad or in international context
- Study more attractive to foreign students
- Improved job opportunities
- Easier entry to Master level

    - Transformers
    - Induction Machines (Asynchronous Machines)

    - Single Phase Machines
